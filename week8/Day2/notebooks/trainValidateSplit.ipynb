```python
import json

INPUT_PATH = "/home/prateek/Prateek/LaunchPad/week8/Day2/data/cleaned.jsonl"
OUTPUT_PATH = "/home/prateek/Prateek/LaunchPad/week8/Day2/data/formatted.jsonl"

def format_sample(sample):
    instruction = sample["instruction"]
    input_text = sample.get("input", "").strip()
    output = sample["output"]

    # Combine instruction + input if input exists
    prompt = f"{instruction}\n{input_text}" if input_text else instruction

    # TinyLlama's native chat template
    formatted_text = (
        f"<|system|>\nYou are a helpful assistant.</s>\n"
        f"<|user|>\n{prompt}</s>\n"
        f"<|assistant|>\n{output}</s>"
    )

    return {"text": formatted_text}

with open(INPUT_PATH, "r") as f, open(OUTPUT_PATH, "w") as out:
    for line in f:
        sample = json.loads(line)
        formatted = format_sample(sample)
        out.write(json.dumps(formatted) + "\n")

print("Formatting done.")
```

```python
from sklearn.model_selection import train_test_split

INPUT_PATH = "/home/prateek/Prateek/LaunchPad/week8/Day2/data/formatted.jsonl"

with open(INPUT_PATH, "r") as f:
    lines = f.readlines()

train_lines, val_lines = train_test_split(lines, test_size=0.1, random_state=42)

with open("/home/prateek/Prateek/LaunchPad/week8/Day2/data/train.jsonl", "w") as f:
    f.writelines(train_lines)

with open("/home/prateek/Prateek/LaunchPad/week8/Day2/data/validate.jsonl", "w") as f:
    f.writelines(val_lines)

print("Train samples:", len(train_lines))
print("Validation samples:", len(val_lines))
```
