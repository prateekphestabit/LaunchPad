```python
## Lora adapters + base model = merged model

from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

BASE_MODEL   = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
ADAPTER_PATH = "/home/prateek/Prateek/LaunchPad/week8/Day3/adapters"
MERGED_PATH  = "/home/prateek/Prateek/LaunchPad/week8/Day3/merged_model"

print("Loading base model...")
base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    torch_dtype=torch.float16,
    device_map="cpu",
)

print("Attaching adapters...")
model = PeftModel.from_pretrained(base_model, ADAPTER_PATH)

print("Merging adapters into base model...")
model = model.merge_and_unload()   # fuses LoRA weights permanently into base

print("Saving merged model...")
model.save_pretrained(MERGED_PATH)
tokenizer = AutoTokenizer.from_pretrained(ADAPTER_PATH)
tokenizer.save_pretrained(MERGED_PATH)

print("Done! Merged model saved.")
```
